```python
import time

arr = list(map(int, input("Enter array elements: ").split()))
key = int(input("Enter element to search: "))

start = time.time()

found = False
for i in range(len(arr)):
    if arr[i] == key:
        found = True
        index = i
        break

end = time.time()

if found:
    print("Element found at index:", index)
else:
    print("Element not found")

print("Execution Time:", end - start, "seconds")
print("Time Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")
```

```python
import time

arr = list(map(int, input("Enter sorted array elements: ").split()))
key = int(input("Enter element to search: "))

start = time.time()

low = 0
high = len(arr) - 1
found = False

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        found = True
        index = mid
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

end = time.time()

if found:
    print("Element found at index:", index)
else:
    print("Element not found")

print("Execution Time:", end - start, "seconds")
print("Time Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(log n)")
print("Worst Case   : O(log n)")
```
